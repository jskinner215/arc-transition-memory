# ARC memory policy, bounded package

Gateway execution requires the competition rerun flag; ordinary execution is a local dry-run. Outside a competition rerun, this notebook writes a labeled commit-stage placeholder so Kaggle can select the notebook for evaluation. The competition gateway must generate the real submission during rerun; this notebook never writes a fallback in that branch. Official artifact generation and hosted compatibility remain unverified. The preserved one-seed public checkpoint scored 0.279093% locally, with zero full-game wins. Planning added no score benefit. This is not a Kaggle score, unseen-game result, competitive-readiness claim or LLM-paper evidence. An isolated child runtime uses only staged hashed wheels; no models, GPU services, personal credentials or downloads are used. Scratch outputs are created outside the package in the operating system temporary directory; preserve the printed path.

Profile: `gateway_dryrun_or_competition_rerun`. The public profile replays 25 familiar games at seed 0 and 500 actions each. The smoke profile uses two MIT toolkit fixtures and eight actions each. The source notebook has no execution outputs.

The notebook host may use Python 3.13. The locked agent runtime still requires installed `/usr/bin/python3.12`, which is probed before creating a pipless child environment. Existing host pip installs only the staged, hash-locked wheels through its documented `--python` option. No interpreter or package-index fallback exists.


```python
import base64, json
PACKAGE = json.loads(base64.b64decode(''))
import base64, hashlib, json, os, platform, subprocess, sys, tempfile
from pathlib import Path
if platform.system() != "Linux" or platform.machine().lower() not in {"x86_64", "amd64"}:
    raise RuntimeError("Bootstrap requires Linux x86_64; staged wheels execute only in verified installed CPython 3.12")
RUN_ROOT = Path(tempfile.mkdtemp(prefix="arc-memory-"))
for relative, item in PACKAGE["files"].items():
    target = RUN_ROOT.joinpath(*relative.split("/"))
    if not target.resolve().is_relative_to(RUN_ROOT.resolve()):
        raise ValueError("embedded path escaped destination")
    data = base64.b64decode(item["base64"], validate=True)
    if hashlib.sha256(data).hexdigest() != item["sha256"]:
        raise ValueError("embedded file hash differs")
    target.parent.mkdir(parents=True, exist_ok=True)
    with target.open("xb") as stream:
        stream.write(data)
print("Memory package unpacked:", RUN_ROOT)
```

```python
"""Kaggle phase handoff; the competition gateway owns all scored output."""

from pathlib import Path
import subprocess
import time
import urllib.error
import urllib.request


GATEWAY_GAMES_URL = "http://gateway:8001/api/games"


def run_checked(*args, **kwargs):
    try:
        return subprocess.run(*args, **kwargs)
    except (subprocess.CalledProcessError, subprocess.TimeoutExpired) as error:
        print("Child process failed:", type(error).__name__)
        for label in ("stdout", "stderr"):
            value = getattr(error, label, None)
            if isinstance(value, bytes):
                value = value.decode("utf-8", errors="replace")
            if value:
                print(label + " (last 16384 characters):\n" + value[-16384:])
        print("No operation is retried. A gateway scorecard may already be closed.")
        raise


def run_postclose_diagnostic(report, *args, **kwargs):
    if report.get("status") != "protocol_completed" or report.get("scorecard_cleanup", {}).get("status") != "closed":
        raise RuntimeError("local diagnostics require a completed protocol and confirmed scorecard closure")
    completed = run_checked(*args, check=False, **kwargs)
    diagnostic = {
        "status": "passed" if completed.returncode == 0 else "diagnostic_failed",
        "exit_code": completed.returncode,
        "stdout": (completed.stdout or "")[-16384:],
        "stderr": (completed.stderr or "")[-16384:],
    }
    if completed.returncode:
        print("Post-closure local diagnostic failed; gateway protocol already completed.")
        print(diagnostic)
    return diagnostic


class _NoRedirect(urllib.request.HTTPRedirectHandler):
    def redirect_request(self, request, stream, code, message, headers, new_url):
        raise urllib.error.HTTPError(request.full_url, code, "gateway redirect refused", headers, stream)


def wait_for_gateway(rerun, *, opener=None, clock=time.monotonic, sleep=time.sleep, timeout=600):
    if not rerun:
        return
    if opener is None:
        opener = urllib.request.build_opener(urllib.request.ProxyHandler({}), _NoRedirect())
    deadline = clock() + timeout
    while True:
        remaining = deadline - clock()
        if remaining <= 0:
            raise TimeoutError("competition gateway did not become ready")
        try:
            with opener.open(GATEWAY_GAMES_URL, timeout=min(10, remaining)) as response:
                if response.status == 200:
                    return
        except (OSError, urllib.error.URLError):
            pass
        remaining = deadline - clock()
        if remaining <= 0:
            raise TimeoutError("competition gateway did not become ready")
        sleep(min(5, remaining))


def write_commit_placeholder(rerun, output="/kaggle/working/submission.parquet", *, dataframe_factory=None):
    if rerun:
        return None
    if dataframe_factory is None:
        import pandas

        dataframe_factory = pandas.DataFrame
    output = Path(output)
    if not output.parent.is_dir():
        raise RuntimeError("Kaggle working directory is required for commit output")
    if output.exists():
        raise FileExistsError("refusing to overwrite an existing submission artifact")
    # Kaggle selects this file for Phase B; it contains no measured result.
    frame = dataframe_factory(
        [["1_0", "1", True, 0.0]],
        columns=["row_id", "game_id", "end_of_game", "score"],
    )
    frame.to_parquet(output, index=False)
    return output


"""Use the verified installed CPython 3.12 without downloading an interpreter."""

import json
from pathlib import Path
import platform


TRUSTED_PYTHON312 = Path("/usr/bin/python3.12")
RUNTIME_PROBE = "import importlib.util,json,platform,sys; print(json.dumps({'implementation':platform.python_implementation(),'version':list(sys.version_info[:3]),'system':platform.system(),'machine':platform.machine(),'venv_available':importlib.util.find_spec('venv') is not None,'is_venv':sys.prefix!=sys.base_prefix}))"


def _verify_python312(value, *, require_venv=False):
    if not isinstance(value, dict) or value.get("implementation") != "CPython" or value.get("version", [])[:2] != [3, 12] or value.get("system") != "Linux" or value.get("machine", "").lower() not in {"x86_64", "amd64"} or value.get("venv_available") is not True:
        raise RuntimeError("installed runtime must be Linux x86_64 CPython 3.12 with venv")
    if require_venv and value.get("is_venv") is not True:
        raise RuntimeError("child interpreter must be an isolated virtual environment")


def prepare_installed_runtime(run_root, child_env, host_python, *, runner):
    run_root = Path(run_root)
    runtime = run_root / "runtime"
    if runtime.exists():
        raise FileExistsError("fresh child runtime directory is required")
    if not TRUSTED_PYTHON312.is_file():
        raise RuntimeError("verified /usr/bin/python3.12 is unavailable; no download fallback")
    options = dict(check=True, cwd=run_root, env=child_env, timeout=30, capture_output=True, text=True)
    probe = runner([str(TRUSTED_PYTHON312), "-I", "-c", RUNTIME_PROBE], **options)
    selected = json.loads(probe.stdout)
    _verify_python312(selected)
    pip_help = runner([str(host_python), "-I", "-m", "pip", "--help"], **options)
    if "--python" not in pip_help.stdout:
        raise RuntimeError("existing host pip lacks --python support; no bootstrap fallback")
    create_options = {**options, "timeout": 60}
    runner([str(TRUSTED_PYTHON312), "-I", "-m", "venv", "--without-pip", "--copies", str(runtime)], **create_options)
    child_python = runtime / "bin/python"
    if not child_python.is_file():
        raise RuntimeError("CPython 3.12 venv did not create its interpreter")
    child_probe = runner([str(child_python), "-I", "-c", RUNTIME_PROBE], **options)
    child = json.loads(child_probe.stdout)
    _verify_python312(child, require_venv=True)
    receipt = {
        "host_python_version": platform.python_version(),
        "selected_interpreter": str(TRUSTED_PYTHON312),
        "selected_runtime": selected, "child_runtime": child,
        "host_pip_python_option_verified": True,
        "pip_bootstrapped_in_child": False, "interpreter_downloaded": False,
    }
    (run_root / "bootstrap-runtime-check.json").write_text(json.dumps(receipt, indent=2) + "\n", encoding="utf-8")
    return child_python, receipt


def offline_install_command(host_python, child_python, wheel_directory, requirements):
    return [str(host_python), "-I", "-m", "pip", "--python", str(child_python), "install",
            "--no-index", "--no-deps", "--only-binary=:all:", "--require-hashes",
            "--no-cache-dir", "--disable-pip-version-check", "--find-links=" + str(wheel_directory),
            "-r", str(requirements)]
```

```python
wheel_candidates = [Path.cwd() / "wheels", Path.cwd().parent / "wheels", Path("/kaggle/input/competitions/arc-prize-2026-arc-agi-3/arc_agi_3_wheels")]
WHEEL_DIR = next((path.resolve() for path in wheel_candidates if path.is_dir()), None)
if WHEEL_DIR is None:
    raise RuntimeError("Stage package wheels beside the notebook or use the reviewed Kaggle wheel mount; no download fallback")
wheel_manifest = json.loads((RUN_ROOT / "wheel-lock.json").read_text())
for item in wheel_manifest:
    data = (WHEEL_DIR / item["path"]).read_bytes()
    if hashlib.sha256(data).hexdigest() != item["sha256"]:
        raise ValueError("offline wheel hash differs")
CHILD_ENV = {key: value for key, value in os.environ.items() if key in {"PATH", "SYSTEMROOT", "WINDIR", "TMP", "TEMP", "TMPDIR", "LD_LIBRARY_PATH", "KAGGLE_IS_COMPETITION_RERUN"}}
CHILD_ENV.update(OPERATION_MODE="offline", ARC_API_KEY="", PYTHON_DOTENV_DISABLED="1", PYTHONDONTWRITEBYTECODE="1", OPENBLAS_NUM_THREADS="1", OMP_NUM_THREADS="1", MKL_NUM_THREADS="1", MPLBACKEND="Agg", MPLCONFIGDIR=str(RUN_ROOT / "matplotlib"), PIP_CONFIG_FILE=os.devnull, PIP_NO_INDEX="1", PIP_DISABLE_PIP_VERSION_CHECK="1", PIP_NO_CACHE_DIR="1")
CHILD_PY, bootstrap_check = prepare_installed_runtime(RUN_ROOT, CHILD_ENV, sys.executable, runner=run_checked)
print(json.dumps(bootstrap_check, indent=2))
install = offline_install_command(sys.executable, CHILD_PY, WHEEL_DIR, RUN_ROOT / "requirements-linux.txt")
run_checked(install, check=True, cwd=RUN_ROOT, env=CHILD_ENV, timeout=120, capture_output=True, text=True)
print("Offline isolated runtime installed from", WHEEL_DIR)
```

```python
rerun = os.environ.get("KAGGLE_IS_COMPETITION_RERUN", "").strip().lower() in {"1", "true"}
preflight = [str(CHILD_PY), "-B", "-c", "import json; from memory_runtime import check_runtime,verify_files; verify_files('.',json.load(open('source-lock.json'))['files']); print(json.dumps(check_runtime('runtime-lock.json')))"]
preflight_result = run_checked(preflight, check=True, cwd=RUN_ROOT, env=CHILD_ENV, timeout=30, capture_output=True, text=True)
runtime_check = json.loads(preflight_result.stdout)
(RUN_ROOT / "runtime-check.json").write_text(json.dumps(runtime_check, indent=2) + "\n", encoding="utf-8")
command = [str(CHILD_PY), "-B", str(RUN_ROOT / "gateway_memory.py"), "--output", str(RUN_ROOT / "gateway-results")]
if rerun:
    command.append("--execute-gateway")
wait_for_gateway(rerun)
run_checked(command, check=True, cwd=RUN_ROOT, env=CHILD_ENV, timeout=30000, capture_output=True, text=True)
report = json.loads((RUN_ROOT / "gateway-results/report.json").read_text())
if report["status"] == "protocol_completed":
    validate_code = "import json,sys; from validate_memory_artifacts import validate_memory_artifacts,REQUIRED_SOURCES; from memory_runtime import verify_files; rows=json.load(open('source-lock.json'))['files']; verify_files('.',rows); expected={row['path']:row['sha256'] for row in rows if row['path'] in REQUIRED_SOURCES}; print(json.dumps(validate_memory_artifacts(sys.argv[1],expected_sources=expected,source_root='.')))"
    diagnostic = run_postclose_diagnostic(report, [str(CHILD_PY), "-B", "-c", validate_code, str(RUN_ROOT / "gateway-results")], cwd=RUN_ROOT, env=CHILD_ENV, timeout=30, capture_output=True, text=True)
    (RUN_ROOT / "gateway-artifact-validation.json").write_text(json.dumps(diagnostic, indent=2) + "\n", encoding="utf-8")
    print(json.dumps({"postclosure_local_diagnostic": diagnostic["status"], "exit_code": diagnostic["exit_code"]}))
print(json.dumps({"status": report["status"], "policy": report["configuration"]["policy"], "competition_submission": False, "hosted_kaggle_validation": False, "runtime_check": str(RUN_ROOT / "runtime-check.json"), "checkpoint_python_match": runtime_check["checkpoint_python_match"], "results": str(RUN_ROOT / "gateway-results")}, indent=2))
print("Official submission artifact schema and generation remain unverified; no placeholder is written.")
```
